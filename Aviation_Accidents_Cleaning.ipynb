### Cleaning and constructing Key Measurables

Injuries and robustness to destruction are a key interest point for the client. Clean and impute relevant columns and then create derived fields that best quantifies what the client wishes to track. **Use commenting or markdown to explain any cleaning assumptions as well as any derived columns you create.**

**Construct metric for fatal/serious injuries**

*Hint:* Estimate the total number of passengers on each flight. The likelihood of serious / fatal injury can be estimated as a fraction from this.

```python

#  CLEANING & DERIVED COLUMNS

injury_cols = ['Total.Fatal.Injuries', 'Total.Serious.Injuries',
               'Total.Minor.Injuries', 'Total.Uninjured']

# Convert injuries to numeric, coerce errors
for col in injury_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# Assumption: if all injury cols are NaN for a record, we CANNOT estimate
# passengers, so we drop those rows for the injury analysis later.
# For now, fill NaN with 0 where at least one injury col is present.
mask_has_any = df[injury_cols].notna().any(axis=1)
for col in injury_cols:
    df[col] = df[col].where(mask_has_any, np.nan).fillna(0)

df['Total_onboard'] = df[injury_cols].sum(axis=1)
df.loc[df['Total_onboard'] == 0, 'Total_onboard'] = np.nan  # invalid denominator

print("Total_onboard stats:")
display(df['Total_onboard'].describe())

print("Records with NaN Total_onboard:", df['Total_onboard'].isna().sum())

# SECTION 4: FATAL / SERIOUS INJURY METRIC
# Assumption: Total passengers ≈ Fatal + Serious + Minor + Uninjured
# Likelihood = (fatal + serious) / total onboard

df['Fatal_or_Serious'] = df['Total.Fatal.Injuries'] + df['Total.Serious.Injuries']
df['Fatal_Serious_Ratio'] = df['Fatal_or_Serious'] / df['Total_onboard']

# Cap ratio at 1.0 (in case of data inconsistencies where sum > onboard)
df['Fatal_Serious_Ratio'] = df['Fatal_Serious_Ratio'].clip(upper=1.0)

# Binary flag: any fatal/serious injury
df['Has_Fatal_or_Serious'] = (df['Fatal_or_Serious'] > 0).astype(int)

print("Fatal_Serious_Ratio statistics (non-null):")
display(df['Fatal_Serious_Ratio'].describe())

print("\nRecords with valid ratio:", df['Fatal_Serious_Ratio'].notna().sum())
```

**Aircraft.Damage**
- identify and execute any cleaning tasks
- construct a derived column tracking whether an aircraft was destroyed or not.

```python
# ============================================================
# SECTION 5: AIRCRAFT.DAMAGE — CLEANING & DERIVED COLUMN
# ============================================================
import numpy as np
import pandas as pd

# --- Step 0: Canonicalize the column name if needed ---
# The raw CSV uses 'Aircraft.damage' (lowercase 'd'), so we standardize it.
if 'Aircraft.Damage' not in df.columns and 'Aircraft.damage' in df.columns:
    df = df.rename(columns={'Aircraft.damage': 'Aircraft.Damage'})

col = 'Aircraft.Damage'

# Safety check
if col not in df.columns:
    candidates = [c for c in df.columns if 'damage' in c.lower()]
    raise KeyError(f"'{col}' not found. Similar columns: {candidates}")

print(f"Using column: {col!r}\n")

# --- Step 1: Inspect BEFORE cleaning ---
print("=== BEFORE CLEANING ===")
print(df[col].value_counts(dropna=False))
print(f"NaN count: {df[col].isna().sum()}")

# --- Step 2: Normalize (strip whitespace, title-case, replace placeholders) ---
df[col] = (df[col]
           .astype(str)
           .str.strip()
           .str.title())

df[col] = df[col].replace({
    'Nan': np.nan, 'NAN': np.nan,
    'None': np.nan, 'NONE': np.nan,
    '': np.nan, ' ': np.nan,
    'Unknown': np.nan, 'UNK': np.nan,
})

# --- Step 3: Keep only valid FAA damage categories ---
valid_categories = ['Destroyed', 'Substantial', 'Minor', 'None']
df[col] = df[col].where(df[col].isin(valid_categories), other=np.nan)

print("\n=== AFTER CLEANING ===")
print(df[col].value_counts(dropna=False))
print(f"NaN count: {df[col].isna().sum()} "
      f"({df[col].isna().mean()*100:.2f}%)")

# --- Step 4: Derived binary flag 'Destroyed' ---
# 1 = destroyed, 0 = not destroyed, NaN = unknown (kept as NaN so we don't
# bias the destruction rate downward by assuming unknowns were "not destroyed")
df['Destroyed'] = np.where(df[col].isna(), np.nan,
                    np.where(df[col] == 'Destroyed', 1.0, 0.0))
df['Destroyed'] = df['Destroyed'].astype(float)

print("\n=== DERIVED COLUMN: 'Destroyed' ===")
print(df['Destroyed'].value_counts(dropna=False))

# --- Step 5: Sanity check ---
known = df['Destroyed'].dropna()
print(f"\nDestruction rate (known cases): {known.mean():.3f} "
      f"({known.mean()*100:.2f}% of accidents destroyed the aircraft)")
print(f"Total known: {len(known):,}  |  Total unknown: {df['Destroyed'].isna().sum():,}")

# Verify mapping is correct
print("\n=== Cross-tab: Aircraft.Damage vs Destroyed ===")
print(pd.crosstab(df[col], df['Destroyed'], dropna=False))
```

### Investigate the *Make* column
- Identify cleaning tasks here
- List cleaning tasks clearly in markdown
- Execute the cleaning tasks
- For your analysis, keep Makes with a reasonable number (you can put the threshold at 50 though lower could work as well)

```python
# ============================================================
# SECTION 6: MAKE COLUMN CLEANING
# ============================================================
# Cleaning tasks (in markdown / comments):
# 1. Strip whitespace and normalize casing (Title Case).
# 2. Fix common variants (e.g. "CESSNA AIRCRAFT CO" -> "Cessna").
# 3. Drop rows where Make is NaN (cannot group).
# 4. Keep only Makes with >= 50 records for statistical robustness.

print("Unique Makes (raw):", df['Make'].nunique())
print("NaN Makes:", df['Make'].isna().sum())

# Normalize
df['Make'] = df['Make'].astype(str).str.strip().str.upper()
df['Make'] = df['Make'].replace({'NAN': np.nan, '': np.nan})

# Consolidate common variants
make_map = {
    'CESSNA AIRCRAFT CO': 'CESSNA',
    'CESSNA AIRCRAFT': 'CESSNA',
    'TEXTRON AVIATION INC': 'TEXTRON AVIATION',
    'PIPER AIRCRAFT INC': 'PIPER',
    'PIPER AIRCRAFT': 'PIPER',
    'BOEING COMPANY': 'BOEING',
    'THE BOEING COMPANY': 'BOEING',
    'AIRBUS INDUSTRIE': 'AIRBUS',
    'AIRBUS SAS': 'AIRBUS',
    'MCDONNELL DOUGLAS': 'MCDONNELL DOUGLAS',
    'MCDONNELL DOUGLAS CORPORATION': 'MCDONNELL DOUGLAS',
    'BEECHCRAFT': 'BEECH',
    'HAWKER BEECHCRAFT': 'BEECH',
    'BOMBARDIER INC': 'BOMBARDIER',
    'EMBRAER': 'EMBRAER',
    'LEARJET': 'LEARJET',
    'GULFSTREAM': 'GULFSTREAM AEROSPACE',
}
df['Make'] = df['Make'].replace(make_map)
df['Make'] = df['Make'].str.title()

# Drop NaN Makes
df = df[df['Make'].notna()].copy()

# Keep Makes with >= 50 records
make_counts = df['Make'].value_counts()
valid_makes = make_counts[make_counts >= 50].index
df = df[df['Make'].isin(valid_makes)].copy()

print("\nRemaining Makes:", df['Make'].nunique())
print(df['Make'].value_counts())
```

### Inspect Model column
- Get rid of any NaNs.
- Inspect the column and counts for each model/make. Are model labels unique to each make?
- If not, create a derived column that is a unique identifier for a given plane type.

```python
# ============================================================
# SECTION 7: MODEL COLUMN
# ============================================================
# Cleaning tasks (markdown):
# 1. Drop rows with NaN Model.
# 2. Normalize casing/whitespace.
# 3. Check whether Model labels are unique across Makes.
# 4. Create a unique identifier 'Make_Model' = "Make Model".
# 5. Keep only Make_Model types with >= 20 records for robust stats.

print("NaN Models before drop:", df['Model'].isna().sum())
df = df[df['Model'].notna()].copy()

# --- Normalize ---
df['Model'] = df['Model'].astype(str).str.strip().str.upper()
df['Model'] = df['Model'].replace({'NAN': np.nan, '': np.nan, 'NONE': np.nan})
df = df[df['Model'].notna()].copy()

# --- Are model labels unique per Make? ---
model_by_make = df.groupby('Model')['Make'].nunique().sort_values(ascending=False)
n_multi = (model_by_make > 1).sum()
print(f"\nModels appearing under multiple Makes: {n_multi}")
print("\nExamples of shared Model names:")
print(model_by_make[model_by_make > 1].head(10))

# --- Derived unique identifier ---
df['Make_Model'] = df['Make'].str.upper() + ' ' + df['Model'].str.upper()

# --- Keep Make_Model with >= 20 records ---
mm_counts = df['Make_Model'].value_counts()
valid_mm = mm_counts[mm_counts >= 20].index
df = df[df['Make_Model'].isin(valid_mm)].copy()

print(f"\nUnique Make_Model types: {df['Make_Model'].nunique()}")
print(f"Rows remaining: {len(df):,}")
```

### Cleaning other columns
- there are other columns containing data that might be related to the outcome of an accident. We list a few here:
- Engine.Type
- Weather.Condition
- Number.of.Engines
- Purpose.of.flight
- Broad.phase.of.flight

Inspect and identify potential cleaning tasks in each of the above columns. Execute those cleaning tasks. 

**Note**: You do not necessarily need to impute or drop NaNs here.

```python
# ============================================================
# SECTION 8: OTHER COLUMNS — CLEANING
# ============================================================

other_cols = ['Engine.Type', 'Weather.Condition', 'Number.of.Engines',
              'Purpose.of.flight', 'Broad.phase.of.flight']

print("=== BEFORE CLEANING ===")
for c in other_cols:
    if c in df.columns:
        print(f"\n--- {c} ---")
        print(df[c].value_counts(dropna=False).head(15))

# Engine.Type 
df['Engine.Type'] = (df['Engine.Type']
                     .astype(str).str.strip().str.title()
                     .replace({'Nan': np.nan, '': np.nan,
                               'Unknown': np.nan, 'Unk': np.nan}))
 #Weather.Condition 
df['Weather.Condition'] = (df['Weather.Condition']
                           .astype(str).str.strip().str.upper()
                           .replace({'NAN': np.nan, '': np.nan,
                                     'UNK': np.nan, 'UNKNOWN': np.nan}))
df.loc[~df['Weather.Condition'].isin(['VMC', 'IMC']), 'Weather.Condition'] = np.nan

# Number.of.Engines 
df['Number.of.Engines'] = pd.to_numeric(df['Number.of.Engines'], errors='coerce')
df['Engine_Count_Cat'] = pd.cut(df['Number.of.Engines'],
                                bins=[0, 1, 2, 3, 4, 100],
                                labels=['1', '2', '3', '4', '5+'])

# Purpose.of.flight 
df['Purpose.of.flight'] = (df['Purpose.of.flight']
                           .astype(str).str.strip().str.title()
                           .replace({'Nan': np.nan, '': np.nan}))
purpose_counts = df['Purpose.of.flight'].value_counts()
rare = purpose_counts[purpose_counts < 50].index
df['Purpose.of.flight'] = df['Purpose.of.flight'].replace({p: 'Other' for p in rare})

# Broad.phase.of.flight 
df['Broad.phase.of.flight'] = (df['Broad.phase.of.flight']
                               .astype(str).str.strip().str.title()
                               .replace({'Nan': np.nan, '': np.nan}))
phase_counts = df['Broad.phase.of.flight'].value_counts()
rare = phase_counts[phase_counts < 50].index
df['Broad.phase.of.flight'] = df['Broad.phase.of.flight'].replace({p: 'Other' for p in rare})

print("\n=== AFTER CLEANING ===")
for c in other_cols:
    if c in df.columns:
        print(f"\n--- {c} ---")
        print(df[c].value_counts(dropna=False).head(15))
```
